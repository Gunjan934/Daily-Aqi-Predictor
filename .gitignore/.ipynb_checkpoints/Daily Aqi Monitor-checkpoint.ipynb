```python
df["Date"]=pd.to_datetime(df["Date"])
city_df=df[df["City"]=="Ahmedabad"].copy()
city_df=city_df.sort_values("Date")
city_df=city_df.set_index("Date")
```

```python
data=city_df[features].copy()
```

```python
data
```

```python
print(city_df.columns.tolist())

print("Duplicate columns:")
print(city_df.columns[city_df.columns.duplicated()])
```

```python
split = int(len(city_df) * 0.8)

train_data = data.iloc[:split].copy()

test_data = data.iloc[split:].copy()

print("Training data:", train_data.shape)
print("Testing data:", test_data.shape)
```

```python
train_data
```

```python
train_medians=train_data[features].median()
train_data[features]=train_data[features].fillna(train_medians)
test_data[features]=test_data[
    
    features].fillna(train_medians)

```

```python
print(train_data.isnull().sum())
print(test_data.isnull().sum())
```

```python
print("Training AQI available:",
      train_data["AQI"].notna().sum())

print("Testing AQI available:",
      test_data["AQI"].notna().sum())
```

```python
X_scaler=MinMaxScaler()
X_train_scaled=X_scaler.fit_transform(
    train_data[features])
X_test_scaled=X_scaler.transform(
    test_data[features]
)
```

```python
y_scaler=MinMaxScaler()
y_train_valid=train_data["AQI"].dropna().values.reshape(-1,1)
y_scaler.fit(y_train_valid)
```

```python
X_scaled = np.vstack([
    X_train_scaled,
    X_test_scaled
])
```

```python
print("X_scaled:", X_scaled.shape)

print("Train features:", X_train_scaled.shape)

print("Test features:", X_test_scaled.shape)
```

```python
window_size=7
```

```python
X_all=timeseries_dataset_from_array(
    data=X_scaled,
    targets=None,
    sequence_length=window_size,
    sequence_stride=1,
    batch_size=None,
    shuffle=False
)
X_all=np.array(list(X_all))
```

```python
y_all=city_df["AQI"].values
target_indices=np.arange(window_size,len(city_df))
```

```python
valid_mask = ~np.isnan(y_all[window_size:])
X = X_all[:-1][valid_mask]
y = y_all[window_size:][valid_mask].reshape(-1, 1)
target_indices = target_indices[valid_mask]
```

```python
print("X_all:", X_all.shape)
print("X:", X.shape)
print("y:", y.shape)
print("Target indices:", target_indices.shape)
```

```python
train_mask = target_indices < split
test_mask = target_indices >= split

X_train = X[train_mask]
y_train = y[train_mask]

X_test = X[test_mask]
y_test = y[test_mask]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
val_split = int(len(X_train) * 0.8)

X_tr = X_train[:val_split]
y_tr = y_train[:val_split]

X_val = X_train[val_split:]
y_val = y_train[val_split:]

print(X_tr.shape)
print(X_val.shape)
```

```python
y_tr_scaled = y_scaler.transform(y_tr)
y_val_scaled = y_scaler.transform(y_val)
```

```python
print(y_tr_scaled.min(), y_tr_scaled.max())
print(y_val_scaled.min(), y_val_scaled.max())
```

```python
from tensorflow.keras.layers import LSTM
model = Sequential([
    Input(shape=(7, 9)),

     LSTM(64, activation="tanh"),

    Dropout(0.2),

    Dense(32, activation="relu"),

    Dense(1, activation="linear")
])
```

```python
model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"])
```

```python
model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)
history = model.fit(
    X_tr,
    y_tr_scaled,
    validation_data=(X_val, y_val_scaled),
    epochs=30,
    batch_size=32,
    shuffle=False,
    callbacks=[early_stop]
)
```

```python
print("y_train minimum:", y_tr_scaled.min())
print("y_train maximum:", y_tr_scaled.max())

print("y_val minimum:", y_val_scaled.min())
print("y_val maximum:", y_val_scaled.max())
```

```python
plt.figure(figsize=(10, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epochs")
plt.ylabel("MSE Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.show()
```

```python
y_pred_scaled = model.predict(X_test)
```

```python
y_pred = y_scaler.inverse_transform(y_pred_scaled)

# y_test is already in original AQI scale
y_actual = y_test
```

```python
print("Actual AQI range:",
      y_actual.min(), y_actual.max())

print("Scaled prediction range:",
      y_pred_scaled.min(), y_pred_scaled.max())

print("Predicted AQI range:",
      y_pred.min(), y_pred.max())
```

```python
print("Minimum AQI:", y_scaler.data_min_)
print("Maximum AQI:", y_scaler.data_max_)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y_actual, y_pred)
mse = mean_squared_error(y_actual, y_pred)
rmse = np.sqrt(mse)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```

# Baseline comparison

```python
# Previous observation's AQI
baseline_pred = city_df["AQI"].shift(1).values

# Actual AQI
baseline_actual = city_df["AQI"].values

# Select only test target positions
baseline_indices = target_indices[test_mask]

baseline_pred = baseline_pred[baseline_indices]
baseline_actual = baseline_actual[baseline_indices]

# Remove observations where previous AQI is missing
valid = ~np.isnan(baseline_pred)

baseline_pred = baseline_pred[valid]
baseline_actual = baseline_actual[valid]
```

```python
baseline_mae = mean_absolute_error(
    baseline_actual,
    baseline_pred
)

baseline_mse = mean_squared_error(
    baseline_actual,
    baseline_pred
)

baseline_rmse = np.sqrt(baseline_mse)

print("Baseline MAE:", baseline_mae)
print("Baseline MSE:", baseline_mse)
print("Baseline RMSE:", baseline_rmse)
```

```python
print("Actual AQI mean:", y_actual.mean())
print("Predicted AQI mean:", y_pred.mean())

print("Actual AQI std:", y_actual.std())
print("Predicted AQI std:", y_pred.std())
```
